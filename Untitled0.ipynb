```python
from google.colab import userdata
import google.generativeai as genai

# Configure Gemini API
genai.configure(api_key=userdata.get("GEMINI_API_KEY"))

# Create Model
model = genai.GenerativeModel(
    model_name="gemini-3.5-flash-lite",
    system_instruction="""
You are a professional social media content writer.

Generate a post according to the platform:

1. LinkedIn:
- Around 120 words
- Professional tone
- 3-5 hashtags
- End with a question

2. Twitter/X:
- Short and concise
- Maximum 280 characters
- 1-2 hashtags

3. Instagram:
- Engaging
- Use emojis
- 5-8 hashtags

4. Facebook:
- Friendly and conversational
- Medium-length post
"""
)

# Function to generate content
def input_info(topic, platform, tone):

    prompt = f"""
Topic: {topic}
Platform: {platform}
Tone: {tone}

Generate a social media caption suitable for the given platform.
"""

    response = model.generate_content(
        prompt,
        generation_config={
            "temperature": 0.9,
            "max_output_tokens": 200
        }
    )

    return response

# Clear previous file
open("social_posts.txt", "w").close()

# Generate 4 posts
for i in range(4):

    print(f"\nPost {i+1}")

    topic = input("Enter the Topic: ")
    platform = input("Enter the Platform (LinkedIn/Twitter/X/Instagram/Facebook): ")
    tone = input("Enter the Tone: ")

    response = input_info(topic, platform, tone)

    print("\nGenerated Post:\n")
    print(response.text)

    with open("social_posts.txt", "a") as file:
        file.write("="*50 + "\n")
        file.write(f"Topic: {topic}\n")
        file.write(f"Platform: {platform}\n")
        file.write(f"Tone: {tone}\n\n")
        file.write("Generated Post:\n")
        file.write(response.text)
        file.write("\n\n")

print("\nAll posts have been saved successfully in 'social_posts.txt'.")
```
